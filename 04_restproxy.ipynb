```python
#| export
from text_generation import Client

client = Client("http://127.0.0.1:999")

def get_sample(prompt, length:int, num_samples:int, allow_linebreak:bool):
    lm_text = '<LM>'+prompt
    result = client.generate(prompt, temperature=.5, repetition_penalty=5.0, typical_p=0.9, 
                       max_new_tokens=length, ).generated_text
    return [result]
```

```python
print(client.generate('<LM>На словах ты Лев Толстой, а на деле', temperature=.5, repetition_penalty=5.0, typical_p=0.9, 
                       max_new_tokens=30, ).generated_text)
```

```python
print(get_sample('На словах ты Лев Толстой, а на деле', 50, 1, False))
```
